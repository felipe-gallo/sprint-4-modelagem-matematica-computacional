## Questão 1 - Energia solar e vetores
### Dados e modelo
A ordem dos vetores é: 8h–10h, 10h–12h, 12h–14h, 14h–16h e 16h–18h.

$p=(0{,}85;0{,}55;0{,}35;0{,}30;0{,}50)$, em R$/kWh.

$q=(16;24;28;26;18)$ e $s=(4;12;25;20;8)$, em kWh.

A energia comprada da rede é $r_i=\max(q_i-s_i,0)$. Neste exercício, $q_i>s_i$ em todos os períodos; portanto, $\mathbf r=\mathbf q-\mathbf s$. A prioridade é usar a energia solar disponível no próprio período.

```python
periodos = ["8h–10h", "10h–12h", "12h–14h", "14h–16h", "16h–18h"]
p = np.array([0.85, 0.55, 0.35, 0.30, 0.50])  # R$/kWh
q = np.array([16, 24, 28, 26, 18])             # kWh
s = np.array([4, 12, 25, 20, 8])              # kWh

r = np.maximum(q - s, 0)
print("Vetor da energia da rede (kWh):", r)
print("Energia total adquirida da rede:", r.sum(), "kWh")
```

### Questão 1(c) - Produto escalar no NumPy
O custo é $C=\mathbf p\cdot\mathbf r=\sum_{i=1}^{5}p_i r_i$. Cada parcela é o preço do período multiplicado pela energia comprada da rede naquele período. A unidade é $(\mathrm{R\$}/\mathrm{kWh})\times\mathrm{kWh}=\mathrm{R\$}$. Não se multiplica novamente por duas horas: os dados de $q$ e $s$ já representam energia em kWh.

```python
custos = p * r
custo_total = np.dot(p, r)

for periodo, energia, custo in zip(periodos, r, custos):
    valor = f"{custo:.2f}".replace(".", ",")
    print(f"{periodo}: {energia} kWh da rede; custo = R$ {valor}")
print("Custo total: R$", f"{custo_total:.2f}".replace(".", ","))
```

### Resultado e interpretação
$$\mathbf r=(12;12;3;6;10)\;\mathrm{kWh}.$$
$$C=0{,}85(12)+0{,}55(12)+0{,}35(3)+0{,}30(6)+0{,}50(10)$$
$$C=10{,}20+6{,}60+1{,}05+1{,}80+5{,}00=\boxed{\mathrm{R\$}\;24{,}65}.$$
A demanda total é de 112 kWh, sendo 69 kWh solares e 43 kWh da rede. A compra complementar atende à demanda em cada intervalo.

## Questão 2 - Potência dos carregadores e sistemas lineares
### Questão 2(a) - Modelagem
Sejam $x$, $y$ e $z$ as quantidades de veículos nas potências de 7 kW, 11 kW e 22 kW, respectivamente. São contagens inteiras não negativas.

$$\begin{cases}x+y+z=6\\y+z=4\\7x+11y+22z=69\end{cases}$$

A primeira equação soma todos os veículos; a segunda conta os quatro veículos nas potências de 11 ou 22 kW; a terceira soma as contribuições de potência, em kW.

### Questão 2(b) - Forma matricial e referência manuscrita
$$A\mathbf u=\mathbf b,\quad A=\begin{pmatrix}1&1&1\\0&1&1\\7&11&22\end{pmatrix},\quad\mathbf u=\begin{pmatrix}x\\y\\z\end{pmatrix},\quad\mathbf b=\begin{pmatrix}6\\4\\69\end{pmatrix}.$$

Na matriz aumentada, $L_3\leftarrow L_3-7L_1$ produz a última linha $(0,4,15\mid27)$. Em seguida, $L_3\leftarrow L_3-4L_2$ produz $(0,0,11\mid11)$.

Por substituição retroativa: $11z=11\Rightarrow z=1$; $y+1=4\Rightarrow y=3$; $x+3+1=6\Rightarrow x=2$.

### Questão 2(c) - Comparação com o solver

```python
A = np.array([[1, 1, 1],
              [0, 1, 1],
              [7, 11, 22]], dtype=float)
b = np.array([6, 4, 69], dtype=float)

solucao_numpy = np.linalg.solve(A, b)
solucao_manual = np.array([2, 3, 1], dtype=float)

print("Solução do NumPy [x, y, z]:", solucao_numpy)
print("Solução por eliminação gaussiana:", solucao_manual)
print("Soluções coincidem:", np.allclose(solucao_numpy, solucao_manual))
for potencia, quantidade in zip([7, 11, 22], solucao_numpy):
    print(f"{quantidade:.0f} veículo(s) na potência de {potencia} kW")
```

### Conferência nas equações originais
A comparação usa tolerância numérica porque o solver opera com números de ponto flutuante.

```python
x, y, z = solucao_numpy
print(f"Total de veículos: {x + y + z:.0f} = 6")
print(f"Veículos a 11 ou 22 kW: {y + z:.0f} = 4")
print(f"Potência total: {7*x + 11*y + 22*z:.0f} kW = 69 kW")
print("Resíduo A @ solução - b:", A @ solucao_numpy - b)

assert np.array_equal(r, [12, 12, 3, 6, 10])
assert np.isclose(custo_total, 24.65)
assert np.array_equal(s + r, q)
assert np.allclose(solucao_numpy, solucao_manual)
assert np.allclose(A @ solucao_numpy, b)
assert np.all(solucao_numpy >= 0)
assert np.allclose(solucao_numpy, np.rint(solucao_numpy))
print("Todas as verificações foram aprovadas.")
```
